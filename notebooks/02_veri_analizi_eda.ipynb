# 📊 Keşifsel Veri Analizi (EDA)

Bu notebook'ta MovieLens veri setini detaylı olarak analiz edeceğiz.

## 📚 Bu Notebook'ta Öğrenecekleriniz

1. Veri setinin yapısı
2. Temel istatistikler
3. Puan dağılımı
4. Film türleri analizi
5. Zaman bazlı analiz

```python
# Kütüphaneleri import et
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
sys.path.insert(0, '..')

# Grafik ayarları
plt.style.use('seaborn')
plt.rcParams['figure.figsize'] = (12, 6)
%matplotlib inline

# Veriyi yükle
from src.data_loader import veri_yukle
filmler, puanlar = veri_yukle()
```

## 1. Veri Yapısı

```python
# Film tablosu bilgileri
print("🎬 FİLM TABLOSU")
print("="*50)
print(f"Satır sayısı: {len(filmler):,}")
print(f"Sütunlar: {list(filmler.columns)}")
print("\nVeri tipleri:")
print(filmler.dtypes)
print("\nİlk 5 satır:")
display(filmler.head())
```

```python
# Puan tablosu bilgileri
print("⭐ PUAN TABLOSU")
print("="*50)
print(f"Satır sayısı: {len(puanlar):,}")
print(f"Sütunlar: {list(puanlar.columns)}")
print("\nTemel istatistikler:")
display(puanlar.describe())
```

## 2. Puan Dağılımı

```python
# Puan dağılımı histogramı
plt.figure(figsize=(10, 6))
puanlar['rating'].hist(bins=10, edgecolor='black')
plt.xlabel('Puan', fontsize=12)
plt.ylabel('Frekans', fontsize=12)
plt.title('Puan Dağılımı', fontsize=14)
plt.xticks([0.5, 1, 1.5, 2, 2.5, 3, 3.5, 4, 4.5, 5])
plt.show()

print(f"\nOrtalama Puan: {puanlar['rating'].mean():.2f}")
print(f"Medyan Puan: {puanlar['rating'].median():.2f}")
```

## 3. Film Türleri Analizi

```python
# Tüm türleri çıkar
tum_turler = filmler['genres'].str.split('|').explode()
tur_sayilari = tum_turler.value_counts()

# Grafik çiz
plt.figure(figsize=(12, 8))
tur_sayilari.plot(kind='barh')
plt.xlabel('Film Sayısı')
plt.ylabel('Tür')
plt.title('Film Türlerinin Dağılımı')
plt.tight_layout()
plt.show()
```

## 4. Kullanıcı ve Film Aktivitesi

```python
# Kullanıcı başına puan sayısı
kullanici_puan_sayisi = puanlar.groupby('userId').size()

print("👤 KULLANICI AKTİVİTESİ")
print("="*40)
print(f"Toplam kullanıcı: {len(kullanici_puan_sayisi):,}")
print(f"Kullanıcı başına ortalama puan: {kullanici_puan_sayisi.mean():.1f}")
print(f"En az puan veren: {kullanici_puan_sayisi.min()}")
print(f"En çok puan veren: {kullanici_puan_sayisi.max()}")
```

```python
# Film başına puan sayısı
film_puan_sayisi = puanlar.groupby('movieId').size()

print("🎬 FİLM POPÜLERLİĞİ")
print("="*40)
print(f"Puanlanan film sayısı: {len(film_puan_sayisi):,}")
print(f"Film başına ortalama puan: {film_puan_sayisi.mean():.1f}")
print(f"En az puanlanan: {film_puan_sayisi.min()} puan")
print(f"En çok puanlanan: {film_puan_sayisi.max()} puan")
```

## 5. Seyreklik Analizi

```python
# Seyreklik hesapla
n_kullanici = puanlar['userId'].nunique()
n_film = puanlar['movieId'].nunique()
n_puan = len(puanlar)

toplam_hucre = n_kullanici * n_film
seyreklik = 1 - (n_puan / toplam_hucre)

print("📊 SEYREKLİK ANALİZİ")
print("="*40)
print(f"Matris boyutu: {n_kullanici:,} × {n_film:,}")
print(f"Toplam hücre: {toplam_hucre:,}")
print(f"Dolu hücre: {n_puan:,}")
print(f"Seyreklik: %{seyreklik*100:.2f}")
```
