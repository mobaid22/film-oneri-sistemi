# 🎓 Model Eğitimi

Bu notebook'ta farklı öneri modellerini eğiteceğiz.

## 📚 Eğitilecek Modeller

1. İçerik Tabanlı Filtreleme
2. Kullanıcı Tabanlı İşbirlikçi Filtreleme
3. Film Tabanlı İşbirlikçi Filtreleme
4. SVD Matris Ayrıştırma
5. Hibrit Model

```python
import sys
sys.path.insert(0, '..')

from src.data_loader import veri_yukle
from src.preprocessing import egitim_test_ayir
from src.models import (
    ContentBasedModel,
    CollaborativeFilteringModel,
    SVDModel,
    HybridModel
)

# Veriyi yükle ve böl
filmler, puanlar = veri_yukle()
egitim, test = egitim_test_ayir(puanlar, test_orani=0.2)
```

## 1. İçerik Tabanlı Model

```python
# İçerik tabanlı model oluştur ve eğit
model_icerik = ContentBasedModel()
model_icerik.egit(filmler, egitim)

# Örnek öneri
oneriler = model_icerik.oner(kullanici_id=1, n=5)
print("\n🎬 Kullanıcı 1 için öneriler:")
for film_id, tahmin in oneriler:
    film = filmler[filmler['movieId'] == film_id]
    if len(film) > 0:
        print(f"  ⭐ {tahmin:.2f} - {film['title'].values[0]}")
```

## 2. İşbirlikçi Filtreleme

```python
# Kullanıcı tabanlı model
model_kullanici = CollaborativeFilteringModel(tur="kullanici")
model_kullanici.egit(filmler, egitim)

# Benzer kullanıcıları bul
benzerler = model_kullanici.benzer_kullanicilari_bul(1, k=5)
print("\n👥 Kullanıcı 1'e benzer kullanıcılar:")
for uid, benz in benzerler:
    print(f"  Kullanıcı {uid}: {benz:.3f}")
```

## 3. SVD Model

```python
# SVD model oluştur ve eğit
model_svd = SVDModel(n_faktor=50, iterasyon=20)
model_svd.egit(filmler, egitim)
```

## 4. Hibrit Model

```python
# Hibrit model oluştur ve eğit
model_hibrit = HybridModel(icerik_agirligi=0.3, isbirligi_agirligi=0.7)
model_hibrit.egit(filmler, egitim)
```
