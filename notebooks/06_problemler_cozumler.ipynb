## 📚 İncelenecek Problemler

1. Cold Start
2. Sparsity (Seyreklik)
3. Scalability (Ölçeklenebilirlik)

```python
import sys
sys.path.insert(0, '..')

from src.data_loader import veri_yukle, kullanici_film_matrisi_olustur
from src.models import HybridModel
from src.preprocessing import egitim_test_ayir

filmler, puanlar = veri_yukle()
```

## 1. Seyreklik (Sparsity) Analizi

```python
# Kullanıcı-Film matrisini oluştur
matris = kullanici_film_matrisi_olustur(puanlar)

# Seyreklik hesapla
import numpy as np
toplam = matris.size
dolu = matris.notna().sum().sum()
seyreklik = 1 - (dolu / toplam)

print(f"Seyreklik: %{seyreklik*100:.2f}")
print(f"\nBu demek ki matrisin %{seyreklik*100:.2f}'si BOŞ!")
```

## 2. Cold Start Analizi

```python
# Hibrit model ile cold start tespiti
egitim, test = egitim_test_ayir(puanlar, test_orani=0.2)

model = HybridModel()
model.egit(filmler, egitim)

# Farklı kullanıcılar için cold start durumu
for uid in [1, 100, 200]:
    durum = model.cold_start_tespit_et(uid)
    print(f"Kullanıcı {uid}: {durum}")
```
