## 1. Öneri Sistemleri Nedir?

Öneri sistemleri, kullanıcılara **ilgilerini çekebilecek** içerikleri otomatik olarak öneren sistemlerdir.

### 🌍 Gerçek Hayat Örnekleri

| Platform | Ne Önerir? | Nasıl? |
|----------|-----------|--------|
| **Netflix** | Film ve dizi | İzleme geçmişi + Benzer kullanıcılar |
| **Spotify** | Şarkı ve playlist | Dinleme alışkanlıkları + Müzik özellikleri |
| **Amazon** | Ürünler | Satın alma geçmişi + "Bunu alanlar şunu da aldı" |
| **YouTube** | Videolar | İzleme geçmişi + Trendler |

### 🎯 Neden Önemli?

- **Netflix**: Kullanıcıların %80'i önerilerden seçim yapıyor
- **Amazon**: Satışların %35'i önerilerden geliyor
- **YouTube**: İzleme süresinin %70'i önerilerden
